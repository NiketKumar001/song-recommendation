```python
new_df = df[df['track_genre'] == 'indian']
```

```python
new_df.head(3)
```

```python
features = ['popularity', 'danceability', 'energy', 'loudness', 'acousticness', 'instrumentalness', 'liveness', 'valence' ]
info = ['track_id', 'artists', 'album_name', 'track_name']
new_df = new_df.drop_duplicates(subset='track_name').reset_index(drop=True)
```

```python
# new_df = new_df[info +  features]
new_df = new_df[info + features]
```

```python
new_df
```

```python
new_df.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
scalar = StandardScaler()
x = new_df[features]
X_scaled = scalar.fit_transform(x)
similarity = cosine_similarity(X_scaled)
```

```python
new_df.iloc[0]
```

```python
import re

def clean_name(name):
    name = re.sub(r'\(.*?\)', '', name)   # (From "Brahmastra") jaisa bracket wala hissa hatao
    name = re.sub(r'\s-\s.*$', '', name)  # " - Lofi Flip" jaisa dash ke baad wala hissa hatao
    return name.strip().lower()
```

```python
def recommend(song_name):
    song_index = new_df.index[new_df['track_name'] == song_name].tolist()
    if song_index == []:
        return 'Song Not Available'

    scores = similarity[song_index[0]]
    sorted_value = sorted(enumerate(scores), key=lambda x : x[1], reverse=True)
    seen = {clean_name(song_name)}
    recommended_songs= []
    for idx, score in sorted_value:
       base = clean_name(new_df.iloc[idx]['track_name'])
       if base in seen:
           continue
       seen.add(base)
       recommended_songs.append(new_df.iloc[idx][['track_name', 'artists']].to_dict())
       if len(recommended_songs) == 5:
           break
    return recommended_songs
    


```
